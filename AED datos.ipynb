# Smart Valenbici
## Carga de datos

```python
import os
import glob
import zipfile
import pandas as pd

carpeta_x = "valenbici"
lista_dataframes = []

i = 0
for ruta_zip in glob.glob(os.path.join(carpeta_x, "*.zip")):
    print(f'{i:03}/954 -> {i*100/954}',end="\r")
    i+=1
    with zipfile.ZipFile(ruta_zip, 'r') as z:
        for nombre_archivo in z.namelist():
            if nombre_archivo.endswith('.csv'):
                with z.open(nombre_archivo) as f:
                    df_temporal = pd.read_csv(f,sep=';')
                    df_temporal['origen_zip'] = os.path.basename(ruta_zip)
                    df_temporal['origen_csv'] = nombre_archivo
                    
                    lista_dataframes.append(df_temporal)


df_total = pd.concat(lista_dataframes, ignore_index=True)
```

```python
df_total.head()
```

```python

df_total['direccion'] = df_total['Direccion'].combine_first(df_total['address'])

df_total['bicis_disponibles'] = df_total['Bicis_disponibles'].combine_first(df_total['available'])

df_total['espacios_libres'] = df_total['Espacios_libres'].combine_first(df_total['free'])

df_total['espacios_totales'] = df_total['Espacios_totales'].combine_first(df_total['total'])

df_total['fecha_actualizacion_final'] = df_total['fecha_actualizacion'].combine_first(df_total['updated_at'])

df_total.head()
```

```python
df_total.drop(
    columns=[
        'Direccion', 'address',
        'Bicis_disponibles', 'available',
        'Espacios_libres', 'free',
        'Espacios_totales', 'total',
        'fecha_actualizacion', 'updated_at'
    ], 
    inplace=True 
)
```

```python
df_total.head()
```

```python
df_total.shape
```

```python
df_total = df_total.dropna(axis=1, how='all')
df_total = df_total.drop(columns=['origen_zip', 'origen_csv'], errors='ignore')
```

```python
df_total['numero'] = df_total['number_'].combine_first(df_total['Numero'])
df_total['activo'] = df_total['Activo'].combine_first(df_total['open'])
df_total = df_total.drop(columns=['number_', 'Numero','open','Activo'], errors='ignore')
```
